# ZeroPoint 1.0: entrenamiento final
Mismo modelo que 1.0b (texto y opciones juntos, "ninguna" en energía cero, el orden de las opciones no influye) con:

1. **Vocabulario de 25 idiomas** (oficiales de la UE con alfabeto latino + catalán, euskera, gallego): el modelo pasa de 470 MB a unos 80 MB y se entrena ya con él.
2. **Opciones difíciles:** en la mitad de las listas, la mitad de las opciones falsas son las más parecidas a la correcta.
3. **Listas largas** de hasta 150 opciones en el entrenamiento, leídas en bloques de 20.
4. **Calibración** con datos separados: temperatura y nivel del "ninguna" ajustados después de entrenar.
5. **Ajuste del cliente** en 200 pasos, con todas las capas o con 8 congeladas.

| Celda | Qué hace | Tiempo aprox. en L4 |
|---|---|---|
| 2 | Entrena la semilla 0 y evalúa todos los tests | ~30 min |
| 3 | Comprobaciones: orden, repetición, listas largas, CPU con 1, 2 y 4 hilos, GPU frente a CPU | ~15 min |
| 4 | Ajuste con tickets (100, 300, 1.000) | ~10 min |
| 5 | Resumen y descarga de `results-1.0c.zip` | 1 min |
| 6 | Opcional: descarga del modelo `s0.pt` (~240 MB) | — |
